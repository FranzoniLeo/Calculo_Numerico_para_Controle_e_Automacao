```python
import numpy as np
import matplotlib.pyplot as plt
v0 = 4 # m/s
h = 1 # m
g = 9.81 # m/s^2



def R(theta):
    return ((v0/g*np.cos(theta))*(v0*np.sin(theta) + np.sqrt(v0**2 * np.sin(theta)**2 + 2*g*h)))

theta_vec = np.linspace(0, np.pi/2, 50)

R_vec = R(theta_vec)
otimizado = theta_vec[np.argmax(R_vec)]
otimizado_graus = 180/np.pi * otimizado
display(otimizado_graus)
display(2.551120 * 180/np.pi)
print(f'Ângulo de lançamento otimizado: {otimizado:.4f} rad')

plt.plot(theta_vec, R_vec)
plt.xlabel('theta (rad)')
plt.ylabel('R (m)')
plt.title('Alcance R em função do ângulo de lançamento theta')

```
