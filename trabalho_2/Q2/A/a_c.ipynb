```python
# A

import numpy as np
import sympy as sp


x = sp.symbols('x')

f = (sp.sqrt(x**2 + 4))/3 + (sp.sqrt((6-x)**2 + 9))/2

df = sp.diff(f, x)
d2f = sp.diff(df, x)

display(f)
display(df)
display(d2f)

# Convertendo para uma função utilizável

f = sp.lambdify(x, f)
df = sp.lambdify(x, df)
d2f = sp.lambdify(x, d2f)
```

```python


def newton_backtracking(xk, eps, a, max_iters, b=0.9):
    k = 0
    while abs(df(xk)) > eps and k < max_iters:
        if d2f(xk) > 0:
            vk = -df(xk)/d2f(xk)
        else:
            vk = -df(xk)

        while f(xk + vk) > (f(xk) + a*df(xk)*vk):
            vk = vk*b
        
        xk = xk + vk
        k+=1
    return [xk, abs(df(xk)), k]

resultado = newton_backtracking(xk=1.0, eps=1e-6, a=0.4, max_iters=100)
xk = resultado[0]
mod_df = resultado[1]
iteracao = resultado[2]
print(f'xk:', xk)
print(f'modulo derivada:', mod_df)
print(f'iteração:', iteracao)



```

```python
# C
# Conferindo com a lei de Snell

theta1 = np.arctan(xk/2)
theta2 = np.arctan((6-xk)/3)
v1 = 3
v2 = 2
esq = np.sin(theta1)/v1
drt = np.sin(theta2)/v2

snell = esq - drt
print(snell)
```
